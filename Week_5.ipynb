Create a calculator, the user will enter two numbers and an operations (+,-,*,/)

Print the appropriate result.

```python
num = int(input("Enter a number:"))
num2 = int(input("Enter another number:"))
op = input("Enter an operation:")

if op == "+":
  print(num+num2)
elif op == "-":
  print(num-num2)
elif op == "*":
  print(num*num2)
elif op == "/":
  print(num/num2)
else:
  print("no")
```

- Create a program go ask the user to enter a phrase.
- Check if the phrase starts with A or a, and print "It starts with A/a"
- Check if the phrase starts with B or b, and print "It starts with B/b"
- In any other case, print "Starts with something else!"

```python
text = input("Enter a phrase:")
if text[0] == "A" or text[0] == "a":
  print("It starts with A.")
elif text[0] == "B" or text[0] == "b":
  print("It starts with B.")
else:
  print("It starts with caca!")
```
